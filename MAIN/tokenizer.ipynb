```python
text = "Aé₹😊"

print("Original text:", text)
print("Unicode code points:")

for char in text:
    print(char, ord(char))

print("\nUTF-8 bytes:")
print(text.encode("utf-8"))

print("\nIndividual byte values:")
print(list(text.encode("utf-8")))
```
